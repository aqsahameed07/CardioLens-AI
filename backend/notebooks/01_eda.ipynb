```python
sns.countplot(data=df, x="target")
plt.title("Heart Disease Target Distribution")
plt.xlabel("Target")
plt.ylabel("Number of Patients")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.histplot(data=df, x="age", kde=True, ax=axes[0, 0])
axes[0, 0].set_title("Age Distribution")

sns.histplot(data=df, x="trestbps", kde=True, ax=axes[0, 1])
axes[0, 1].set_title("Resting Blood Pressure")

sns.histplot(data=df, x="chol", kde=True, ax=axes[1, 0])
axes[1, 0].set_title("Cholesterol Distribution")

sns.histplot(data=df, x="thalach", kde=True, ax=axes[1, 1])
axes[1, 1].set_title("Maximum Heart Rate")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Matrix")
plt.show()
```

```python
sns.boxplot(data=df, x="target", y="age")
plt.title("Age vs Heart Disease")
plt.show()
```

```python
sns.boxplot(data=df, x="target", y="chol")
plt.title("Cholesterol vs Heart Disease")
plt.show()
```

```python
sns.boxplot(data=df, x="target", y="thalach")
plt.title("Maximum Heart Rate vs Heart Disease")
plt.show()
```
